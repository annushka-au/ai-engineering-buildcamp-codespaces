```python
from dotenv import dotenv_values, find_dotenv

env_file = find_dotenv(usecwd=True)
values = dotenv_values(env_file)

print("Found:", env_file)
print("Variables:", list(values.keys()))
print("OPENAI_API_KEY present:", "OPENAI_API_KEY" in values)
print("OPENAI_API_KEY has value:", bool(values.get("OPENAI_API_KEY")))
```
