```python
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv(override=True)

openai_client = OpenAI()
```

```python
# Send a prompt to the OpenAI Responses API and print the generated text

response = openai_client.responses.create(

    model="gpt-4o-mini",    # Model used to generate the response

    input="Write a short bedtime story about a unicorn"     # Prompt sent to the model
)

print(response.output_text)     # output_text: generated text returned by the model
```
