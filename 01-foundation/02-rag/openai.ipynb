```python
messages = [
    {"role": "user", "content": "tell me about Australia in ten sentences"}
]

response = openai_client.responses.create(
    model='gpt-4o-mini',
    input=messages
)
usage = response.usage
```

```python
# the price dictionary has been copied 25 of September 2026 from https://aishippinglabs.com/courses/ai-buildcamp/foundation/rag-search/openai-api
MODEL_PRICES = {
    "gpt-4o-mini": {"input": 0.15, "output": 0.60},
    "gpt-5-nano": {"input": 0.075, "output": 0.30},
    "gpt-5-mini": {"input": 0.25, "output": 2.00},
    "gpt-5.2": {"input": 1.75, "output": 14.00},
    "gpt-5.2-pro": {"input": 21.00, "output": 168.00},
}
```

```python
def calculate_cost(model_name: str, input_tokens: int, output_tokens: int) -> float:
    prices = MODEL_PRICES[model_name.lower()]
    input_cost = (input_tokens / 1_000_000) * prices["input"]
    output_cost = (output_tokens / 1_000_000) * prices["output"]
    return input_cost + output_cost



cost = calculate_cost(
    model_name="gpt-4o-mini",
    input_tokens=usage.input_tokens,
    output_tokens=usage.output_tokens
)

print(f"Cost: ${cost:.6f}")    
```

```python
instructions = """
Write a short bedtime story about a unicorn.

Use emojis throughout the story: add an appropriate emoji whenever you mention a creature, place, action, or object.

Return the story as HTML.
Highlight important nouns using:
<span style="background-color: yellow;">noun</span>

Highlight key action verbs using:
<span style="background-color: #E6CCFF;">verb</span>

Only highlight the main nouns and the key verbs, not every word.
"""
user_prompt = "Write a short bedtime story about a unicorn"

messages = [
    {"role": "system", "content": instructions},
    {"role": "user", "content": user_prompt},
]

stream = openai_client.responses.create (
    model = 'gpt-4o-mini',
    input = messages,
    stream = True
)

events = []

response = None 
for event in stream: 
    events.append(event)
    if hasattr(event, 'delta'):
        print(event.delta, end = '')
    if hasattr(event, 'response'): 
        response = event.response

#print(response.output_text)

display(HTML(response.output_text))
```
